```python
import pandas as pd
import sqlite3

# ── 1. OXUMA ─────────────────────────────────────────────
df = pd.read_csv("Sample - Superstore.csv", encoding="latin-1")
print(f"Həcm: {df.shape}")

# ── 2. TƏMİZLƏMƏ ─────────────────────────────────────────
# Sütun adlarını normallaşdır
df.columns = (df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_"))

# Tarixləri çevir
df["order_date"] = pd.to_datetime(df["order_date"], dayfirst=False)
df["ship_date"]  = pd.to_datetime(df["ship_date"],  dayfirst=False)

# Yeni sütunlar əlavə et
df["order_year"]    = df["order_date"].dt.year
df["order_month"]   = df["order_date"].dt.month
df["order_quarter"] = df["order_date"].dt.quarter
df["ship_days"]     = (df["ship_date"] - df["order_date"]).dt.days
df["profit_margin"] = (df["profit"] / df["sales"] * 100).round(2)

# Dublikatları sil
evvel = len(df)
df = df.drop_duplicates(subset=["order_id","product_id"])
print(f"Dublikat silindi: {evvel - len(df)}")

# ── 3. BAZAYA YAZ ─────────────────────────────────────────
conn = sqlite3.connect("superstore.db")
df.to_sql("orders", conn, if_exists="replace", index=False)
conn.close()

print(f"\n✅ {len(df)} sətir superstore.db-yə yazıldı")
print(f"\nSütunlar: {list(df.columns)}")
print(f"\nMəlumat nümunəsi:")
print(df[["order_date","customer_name","category","sales","profit","profit_margin"]].head())
```
